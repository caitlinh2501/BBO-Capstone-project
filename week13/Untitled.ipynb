```python
import os
import numpy as np

# ============================================================
# FUNCTION 8 - BUILD WEEK 13 CUMULATIVE DATA
# Run from inside the week13/ folder
# ============================================================

def build_function8_week13():

    # --------------------------------------------------------
    # Load cumulative data from Week 12
    # --------------------------------------------------------
    X_prev = np.load(
        "../week12/function8/initial_inputs.npy"
    )

    Y_prev = np.load(
        "../week12/function8/initial_outputs.npy"
    ).reshape(-1)

    # --------------------------------------------------------
    # Week 12 submitted query + source-of-truth result
    # --------------------------------------------------------
    week12_x = np.array([[
        0.088567,
        0.167731,
        0.131357,
        0.163034,
        0.800372,
        0.500019,
        0.206649,
        0.588480
    ]])

    week12_y = np.array([
        9.999147016163
    ])

    # --------------------------------------------------------
    # Append Week 12 observation
    # --------------------------------------------------------
    X_new = np.vstack([
        X_prev,
        week12_x
    ])

    Y_new = np.concatenate([
        Y_prev,
        week12_y
    ])

    # --------------------------------------------------------
    # Save cumulative Week 13 data
    # --------------------------------------------------------
    os.makedirs(
        "function8",
        exist_ok=True
    )

    np.save(
        "function8/initial_inputs.npy",
        X_new
    )

    np.save(
        "function8/initial_outputs.npy",
        Y_new
    )

    # --------------------------------------------------------
    # Verification
    # --------------------------------------------------------
    best_idx = np.argmax(Y_new)

    print("================================")
    print("FUNCTION 8 - WEEK 13 DATA")
    print("================================")

    print("\nX shape:", X_new.shape)
    print("Y shape:", Y_new.shape)

    print("\nAppended Week 12 input:")
    print(X_new[-1])

    print("\nAppended Week 12 output:")
    print(Y_new[-1])

    print("\nCurrent best input:")
    print(X_new[best_idx])

    print("\nCurrent best output:")
    print(Y_new[best_idx])

    return X_new, Y_new


# Run
X8, Y8 = build_function8_week13()
```
