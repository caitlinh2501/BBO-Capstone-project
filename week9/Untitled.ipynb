```python
import numpy as np
import os

# ============================================================
# FUNCTION 8 - BUILD WEEK 10 DATA
# Run from inside the week9/ folder
# ============================================================

# Existing cumulative data through Week 8
X = np.load("function8/initial_inputs.npy")
Y = np.load("function8/initial_outputs.npy").reshape(-1)

print("Before adding Week 9:")
print("X shape:", X.shape)
print("Y shape:", Y.shape)

# ------------------------------------------------------------
# Week 9 result
# ------------------------------------------------------------

week9_x = np.array([
    0.117062,
    0.183135,
    0.104304,
    0.120414,
    0.770644,
    0.497652,
    0.179588,
    0.558852
])

week9_y = 9.9940246476926

# Append exactly once
X_week10 = np.vstack([
    X,
    week9_x
])

Y_week10 = np.append(
    Y,
    week9_y
)

# ------------------------------------------------------------
# Save Week 10 cumulative dataset
# ------------------------------------------------------------

save_dir = "../week10/function8"
os.makedirs(save_dir, exist_ok=True)

np.save(
    f"{save_dir}/initial_inputs.npy",
    X_week10
)

np.save(
    f"{save_dir}/initial_outputs.npy",
    Y_week10
)

# ------------------------------------------------------------
# Verification
# ------------------------------------------------------------

print("\nAfter adding Week 9:")
print("X shape:", X_week10.shape)
print("Y shape:", Y_week10.shape)

print("\nLast input:")
print(X_week10[-1])

print("\nLast output:")
print(Y_week10[-1])

best_idx = np.argmax(Y_week10)

print("\nCurrent best:")
print(
    X_week10[best_idx],
    "->",
    Y_week10[best_idx]
)

print("\nY range:")
print("min =", Y_week10.min())
print("max =", Y_week10.max())
```
